# 03 — Outlier Analysis
This notebook uses the IQR method to flag unusual Quantity and LineValue observations in eligible positive sales. It does not automatically treat an outlier as an error.

## Learning objectives
- Understand NumPy percentiles, Q1, Q3, and the IQR
- Create separate outlier flags
- Compare the influence of wholesale orders on totals

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from retail_analysis.data_loader import load_retail_data
from retail_analysis.cleaning import clean_transactions, split_transaction_sets
from retail_analysis.features import add_transaction_features
from retail_analysis.outliers import add_outlier_flags
from retail_analysis.analysis import outlier_influence

cleaned, _ = clean_transactions(load_retail_data())
sales, _, _ = split_transaction_sets(cleaned)
sales = add_transaction_features(sales)
sales, thresholds = add_outlier_flags(sales)
thresholds
```

`np.percentile(values, [25, 75])` returns Q1 and Q3. The IQR is `Q3 - Q1`, and values outside the calculated bounds are flagged. If the IQR is zero, `ZeroIQR` indicates that the middle 50% of the values are identical.

```python
# Examine the percentile operation with a small sample
sample = np.array([1, 2, 2, 3, 20])
np.percentile(sample, [25, 75])
```

```python
outlier_influence(sales)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
sns.boxplot(y=sales['Quantity'], ax=axes[0])
sns.boxplot(y=sales['LineValue'], ax=axes[1])
axes[0].set_title('Quantity'); axes[1].set_title('Line value (GBP)')
plt.tight_layout(); plt.show()
```
